```python
import pygame
```

```python
pygame.init()
clock = pygame.time.Clock()
```

```python
display = pygame.display.set_mode((800, 600))
pygame.display.set_caption(title = "Assignment 1: Gabriel Maldonado")
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT or (event.type == pygame.KEYDOWN and event.key == 27):
            running = False
    display.fill((255, 255, 50))
    pygame.display.flip()
    clock.tick(30)
pygame.quit()
```
